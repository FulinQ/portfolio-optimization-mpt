```python
import pandas as pd
```

```python
file_path = "Fund NAV History.xlsx"
xls = pd.ExcelFile(file_path)

# 'Summary' sheet has its header on row 4 (index 3); every fund sheet has
# a title row + blank row before the header on row 3 (index 2).
dfs = {}
for sheet in xls.sheet_names:
    header_row = 3 if sheet == "Summary" else 2
    dfs[str(sheet).replace(' ', '-')] = pd.read_excel(xls, sheet_name=sheet, header=header_row)

dfs.keys()
```

```python
fund_names = [i for i in dfs.keys()][1:]
fund_names
```

```python
summary = dfs['Summary'].dropna(subset=['#']).astype({'#': 'int'}).set_index('#')
summary['Fund'] = summary['Fund'].str.replace(' ', '-')
df_order = summary[['Fund', 'First date']]
df_order = df_order.set_index('First date').sort_index()
print(df_order.head(3))

pivot_order = [i for i in df_order['Fund']]
pivot_order
```
